```python
import random
def jugar_piedra():
    jugador_nombre = input("Escribe tu nombre aquí").capitalize()
    print("Bienvenido al juego", jugador_nombre)

    elemento = ('Piedra', 'Papel', 'Tijera')
    jugador2 = random.choice(elemento)

    vida = 3
    while vida != 6 and vida != 0:
        jugador2 = random.choice(elemento)
        jugador1 = input('Escoge tu elemento : Piedra, Papel o Tijera').capitalize()
        print("Has escogido:", jugador1)
        if jugador1 == 'Piedra':
            if jugador2 == 'Tijera':
                vida += 1
                print('Has ganado un punto')
            elif  jugador2 == 'Papel':
                vida -= 1
                print('Has perdido la ronda')
            else:
                print('Has empatado')


        elif jugador1 == 'Papel':
            if jugador2 == 'Piedra':
                vida += 1
                print('Has ganado un punto')
            elif jugador2 == 'Tijera':
                vida -= 1
                print('Has perdido la ronda')
            else:
                print('Has empatado')
   
        elif jugador1 == 'Tijera':
            if jugador2 == 'Papel':
                vida += 1
                print('Has ganado un punto')
            elif jugador2 == 'Piedra':
                vida -= 1
                print('Has perdido la ronda')
            else:
                print('Has empatado')
        else:
            print("La palabra introducida no es válida")

        print("El ordenador ha escogido:", jugador2)  

    if vida == 6:
        print("Te has pasado el juego")
    elif vida == 0:
        print("Game over")
```

```python
jugar_piedra()
```
