```python
import random


class Juegos:
    def jugar_piedra(self):
        jugador_nombre = input("Escribe tu nombre aquí: ").capitalize()
        print("\nBienvenido al juego", jugador_nombre, "\n")
        elemento = ('Piedra', 'Papel', 'Tijera')
        vida = 3
        while vida != 6 and vida != 0:
            jugador2 = random.choice(elemento)
            jugador1 = input('Escoge tu elemento: Piedra, Papel o Tijera: ').capitalize()
            print("\nHas escogido:", jugador1)
            if jugador1 == 'Piedra':
                if jugador2 == 'Tijera':
                    vida += 1
                    print('Has ganado un punto\n')
                elif jugador2 == 'Papel':
                    vida -= 1
                    print('Has perdido la ronda\n')
                else:
                    print('Has empatado\n')
            elif jugador1 == 'Papel':
                if jugador2 == 'Piedra':
                    vida += 1
                    print('Has ganado un punto\n')
                elif jugador2 == 'Tijera':
                    vida -= 1
                    print('Has perdido la ronda\n')
                else:
                    print('Has empatado\n')
            elif jugador1 == 'Tijera':
                if jugador2 == 'Papel':
                    vida += 1
                    print('Has ganado un punto\n')
                elif jugador2 == 'Piedra':
                    vida -= 1
                    print('Has perdido la ronda\n')
                else:
                    print('Has empatado\n')
            else:
                print("La palabra introducida no es válida\n")
            print("El ordenador ha escogido:", jugador2)
            print("Vidas actuales:", vida, "\n")
        if vida == 6:
            print("🎉 Te has pasado el juego 🎉\n")
        elif vida == 0:
            print("💀 Game over 💀\n")
    def jugar_geografia(self):
        print("Has elegido jugar a Preguntas y Respuestas")
        
        preguntas = {"¿Cuál es el río más largo del mundo?": "Amazonas",
        "¿Cuál es la capital de Australia?": "Canberra",
        "¿En qué continente está Egipto?": "Africa",
        "¿Qué montaña separa Europa de Asia?": "Urales",
        "¿Cuál es el país más grande del mundo?": "Rusia",
        "¿Qué océano baña las costas de Chile?": "Pacifico",
        "¿Cuál es la capital de Canadá?": "Ottawa",
        "¿En qué ciudad está el pico más alto de la Península Ibérica?": "Granada",
        "¿Qué país tiene forma de bota?": "Italia",
        "¿Cuál es el desierto más grande del mundo?": "Sahara"}
        preguntas_correctas = 0
        intentos_fallidos = 0
        max_intentos = 3
        objetivo = 5
        print("\n¡Bienvenido al Juego de Geografía!")
        print(f"\nDebes responder {objetivo} preguntas correctamente para ganar.")
        print(f"\nPero cuidado, ¡solo puedes fallar {max_intentos} veces!\n")
        lista_preguntas = list(preguntas.items())
        random.shuffle(lista_preguntas)
        for pregunta, respuesta_correcta in lista_preguntas:
            if preguntas_correctas >= objetivo:
                if preguntas_correctas >= objetivo:
                    print(f"\n🎉¡Felicidades!🎉 Ganaste con {preguntas_correctas} aciertos.")
                    return
            if intentos_fallidos >= max_intentos:
                print("\n💀¡Se te acabaron los intentos!💀 Perdiste el juego.")
                print(f"Conseguiste {preguntas_correctas} respuestas correctas.")
                return
            print(f"Pregunta: {pregunta}")
            respuesta_usuario = input("Tu respuesta: ").strip().lower()
            if respuesta_usuario == respuesta_correcta.lower():
                preguntas_correctas += 1
                print(f"\n¡Correcto! ✅ Llevas {preguntas_correctas} aciertos.")
                
            else:
                intentos_fallidos += 1
                print(f"\n❌Incorrecto. La respuesta correcta era: {respuesta_correcta}")
                print(f"Te quedan {max_intentos - intentos_fallidos} intentos restantes.")
        print("\n¡Se acabaron las preguntas!No alcanzaste los {objetivo} aciertos. Obtuviste {preguntas_correctas}.")
        
            
        
    def jugar_ahorcado(self):
        estados = ahorcado = [
    '''
    ''',
    '''
       ------
       |    |
       |
       |
       |
       |
    ''',
    '''
       ------
       |    |
       |    O
       |
       |
       |
    ''',
    '''
       ------
       |    |
       |    O
       |    |
       |
       |
    ''',
    '''
       ------
       |    |
       |    O
       |   /|
       |
       |
    ''',
    '''
       ------
       |    |
       |    O
       |   /|\\
       |
       |
    ''',
    '''
       ------
       |    |
       |    O
       |   /|\\
       |   /
       |
    ''',
    '''
       ------
       |    |
       |    O
       |   /|\\
       |   / \\
       |
    '''
]
        lista_palabras = ["cereza", "limon", "piña", "mango"]
        palabra_secreta = random.choice(lista_palabras)
        random.shuffle(lista_palabras)
        palabra_oculta = ["_"] * len(palabra_secreta)
        letras_usadas = []
        intentos = 7
        print("¡Bienvenida al juego El Ahorcado!")
        print("Pista: la palabra secreta es una fruta")
        print("Adivina la palabra, tienes 7 intentos\n")
        while intentos > 0 and "_" in palabra_oculta:
            print(estados[7 - intentos])
            print("\nPalabra:", " ".join(palabra_oculta))
            print("Letras usadas:", ", ".join(letras_usadas))
            print(f"Intentos restantes: {intentos}\n")
            letra = input("Elige una letra: ").lower()
            print()
            if letra in letras_usadas:
                print("Ya has usado esa letra antes, intenta con otra.\n")
                continue
            letras_usadas.append(letra)
            if letra in palabra_secreta:
                for i in range(len(palabra_secreta)):
                    if palabra_secreta[i] == letra:
                        palabra_oculta[i] = letra
                print("¡Correcto!\n")
            else:
                intentos -= 1
                print("Incorrecto, esa letra no está en la palabra.\n")
                print(estados[7 - intentos])
        if "_" not in palabra_oculta:
            print("🎉¡Felicidades!🎉 Has adivinado la palabra:", palabra_secreta)
        else:
            print("💀Has perdido.💀 La palabra secreta era:", palabra_secreta)


```
